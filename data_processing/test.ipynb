```python
from pathlib import Path
import pandas as pd

# Set the folder that contains your CSV files.
folder_path = Path(r"C:\Users\user\Desktop\Gibeom\1. HI Lab\0. Projects\0. On going\1. [Lead] Sanitizing DEG w Jiwon\1. 실험자료\260819\0minimized_csv파일\DC peak")
```

```python
# If your CSV files have a header row, set this to True.
has_header = False

# Search CSV files in subfolders too.
search_subfolders = True

# Output folder name.
output_folder_name = "filtered_csv"

if not folder_path.exists():
    raise FileNotFoundError(f"Folder not found: {folder_path}")

if not folder_path.is_dir():
    raise NotADirectoryError(f"Not a folder: {folder_path}")

output_folder = folder_path / output_folder_name
output_folder.mkdir(parents=True, exist_ok=True)

search_iter = folder_path.rglob("*") if search_subfolders else folder_path.glob("*")

csv_files = sorted(
    path for path in search_iter
    if path.is_file()
    and path.suffix.lower() == ".csv"
    and output_folder not in path.parents
)

print(f"Folder: {folder_path}")
print(f"CSV found: {len(csv_files)}")
for path in csv_files[:10]:
    print(f" - {path}")

if not csv_files:
    raise FileNotFoundError(
        f"No CSV files found in: {folder_path}\n"
        "Check 1) the folder path 2) whether files are inside subfolders 3) whether the extension is .csv or .CSV"
    )

saved_count = 0

for csv_path in csv_files:
    df = pd.read_csv(csv_path, header=0 if has_header else None)

    if df.shape[1] < 2:
        print(f"Skipped (less than 2 columns): {csv_path.name}")
        continue

    df_filtered = df.iloc[:, :2]
    relative_path = csv_path.relative_to(folder_path)
    save_path = output_folder / relative_path
    save_path.parent.mkdir(parents=True, exist_ok=True)
    df_filtered.to_csv(save_path, index=False, header=has_header)
    saved_count += 1
    print(f"Saved: {save_path}")

print(f"Done. Saved {saved_count} file(s) to: {output_folder}")
```
